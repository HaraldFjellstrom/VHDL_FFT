### **Hardware Verification Report: 16-bit Block Floating Point FFT/IFFT Core**

This report summarizes the functional verification and performance analysis of a high-performance **Radix-2 FFT/IFFT processor** implemented in VHDL. To balance high dynamic range with hardware efficiency, the core utilizes a **Block Floating Point (BFP)** architecture with **convergent rounding** and automated stage-scaling logic. 

The design was subjected to a comprehensive regression suite covering edge-case impulses, white noise, and small-signal "torture" tests across transform lengths from $N=16$ to $N=2048$. Results demonstrate a peak system precision of **14.66 ENOB** and stable performance in a full-scale IFFT round-trip, confirming the core is fully verified and ready for adaptation into hardware.

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.utils import capture
from fft_utils import run_sim, analyze_fft_performance

# ---------------------------------------------------------
# 1. Signal Generation Library
# ---------------------------------------------------------
def gen_impulse(N):
    sig = np.zeros(N, dtype=complex)
    sig[0] = 0.99  # Near full scale impulse at t=0
    return sig

def gen_dc_nyquist(N):
    # DC component + Nyquist frequency component
    t = np.arange(N)
    return 0.5 * np.ones(N) + 0.5 * np.cos(np.pi * t)

def gen_white_noise(N):
    # Full scale random noise
    return np.random.uniform(-0.9, 0.9, N) + 1j*np.random.uniform(-0.9, 0.9, N)

def gen_linear_chirp(N):
    t = np.arange(N) / N
    # Sweeps from 0 to Nyquist
    phi = np.pi * (N/2) * t**2
    return 0.99 * np.exp(1j * phi)

def gen_pi_4_stress(N):
    t = np.arange(N)
    # Target Bin N/8 to force 45-degree twiddle multiplications
    freq_bin = N // 8
    return 0.99 * np.exp(1j * 2 * np.pi * freq_bin * t / N)

def gen_damped_sine(N, scale=0.99):
    """Mimics a real-world transient with multiple frequencies."""
    t = np.arange(N) / N # Normalized time
    decay = np.exp(-5.0 * t) # Decay envelope
    
    # Multi-tone core
    freq1, freq2 = 5.3, 20.7 
    sig = (np.sin(2 * np.pi * freq1 * t) + 0.5 * np.sin(2 * np.pi * freq2 * t)) / 1.5
    
    return scale * decay * sig

def gen_torture_test(N, db_level=-80):
    """The 'Goliath vs David' test: One full-scale tone, one tiny tone."""
    t = np.arange(N)
    goliath = 0.95 * np.exp(1j * 2 * np.pi * 15 * t / N) # Bin 15
    amp_david = 10**(db_level/20)
    david = amp_david * np.exp(1j * 2 * np.pi * 100 * t / N) # Bin 100
    return goliath + david

# ---------------------------------------------------------
# 2. Configuration & Execution
# ---------------------------------------------------------
N_values = [16, 32, 64, 128, 256, 512, 1024, 2048]
test_scenarios = [
    {"name": "Dirac Impulse", "func": gen_impulse},
    {"name": "DC & Nyquist", "func": gen_dc_nyquist},
    {"name": "White Noise",  "func": gen_white_noise},
    {"name": "Linear Chirp", "func": gen_linear_chirp},
    {"name": "Pi/4 Stress",  "func": gen_pi_4_stress},
    {"name": "Damped Sine (Full)",  "func": lambda N: gen_damped_sine(N, scale=0.99)},
    {"name": "Damped Sine (-20dB)", "func": lambda N: gen_damped_sine(N, scale=0.10)},
    {"name": "Damped Sine (-40dB)", "func": lambda N: gen_damped_sine(N, scale=0.01)},
    {"name": "Torture (-80dB)", "func": lambda N: gen_torture_test(N, -80)},
    {"name": "Torture (-95dB)", "func": lambda N: gen_torture_test(N, -95)},
]

# Results storage: {test_name: [mse_values_per_N]}
regression_results = {test["name"]: {'N': [], 'sqnr': [], 'max_err': [], 'exp': [], 'enob': [], 'cycles': []} for test in test_scenarios}

print("🚀 Starting FFT Regression Suite...")

for test in test_scenarios:
    print(f"\n--- Scenario: {test['name']} ---")
    
    # 1. Generate a representative version for visualization (N=256)
    N_repr = 256
    repr_signal = test["func"](N_repr).astype(complex)

    # 2. Plot the Stimulus once per Scenario
    plt.figure(figsize=(12, 3))
    plt.plot(repr_signal.real, label="Real (In-phase)", color='dodgerblue', alpha=0.8)
    plt.plot(repr_signal.imag, label="Imag (Quadrature)", color='coral', alpha=0.6)
    plt.title(f"Representative Input Stimulus: {test['name']} (Shown at N={N_repr})")
    plt.xlabel("Sample Index")
    plt.ylabel("Normalized Amplitude")
    plt.legend(loc="upper right")
    plt.grid(True, alpha=0.2)
    plt.tight_layout()
    plt.show()
    
    for N in N_values:
        # Generate the specific corner-case signal
        input_signal = test["func"](N).astype(complex)

        with capture.capture_output() as cap:
            try:
                hw_sig, hw_exp, cycles = run_sim(input_signal, SCALE_THRESHOLD=(8191))
                sim_success = True
            except Exception as e:
                sim_success = False
                error_msg = e
        
        if sim_success:
            stats = analyze_fft_performance(input_signal, hw_sig, hw_exp)
            
            # Calculate ENOB: (SQNR - 1.76) / 6.02
            current_enob = (stats['sqnr_db'] - 1.76) / 6.02
            current_enob = min(current_enob, 16.0)
            
            regression_results[test['name']]['N'].append(N)
            regression_results[test['name']]['sqnr'].append(stats['sqnr_db'])
            regression_results[test['name']]['max_err'].append(stats['max_diff'])
            regression_results[test['name']]['exp'].append(hw_exp)
            regression_results[test['name']]['enob'].append(current_enob)
            regression_results[test['name']]['cycles'].append(cycles)
            
            print(f"  ✅ N={N:<4} | SQNR: {stats['sqnr_db']:.2f} dB | Max Err: {stats['max_diff']:.2e} | Exp: {hw_exp} | Enob: {current_enob:.2f}")
        else:
            print(f"  ❌ N={N:<4} | FAILED")
            regression_results[test['name']].append(np.nan)

# ---------------------------------------------------------
# 3. Regression Plotting
# ---------------------------------------------------------
print("\n📊 Generating Multi-Metric Analysis Dashboard...")

# Create 3 subplots sharing the same X-axis (N)
fig, (ax_sqnr, ax_err, ax_exp, ax_cc) = plt.subplots(4, 1, figsize=(12, 14), sharex=True)
colors = plt.cm.tab10(np.linspace(0, 1, len(test_scenarios)))

for i, (test_name, data) in enumerate(regression_results.items()):
    # 1. SQNR Plot (Higher is better)
    ax_sqnr.plot(data['N'], data['sqnr'], marker='o', label=test_name, color=colors[i], linewidth=2)
    
    # 2. Max Error Plot (Lower is better - Log scale)
    ax_err.plot(data['N'], data['max_err'], marker='s', label=test_name, color=colors[i], linewidth=2, linestyle='--')
    
    # 3. Block Exponent Plot (Hardware Activity)
    ax_exp.step(data['N'], data['exp'], where='mid', marker='x', label=test_name, color=colors[i], alpha=0.8)

    # 4. Clock Cycles Plot
    ax_cc.step(data['N'], data['cycles'], where='mid', marker='x', label=test_name, color=colors[i], alpha=0.8)

# Formatting Top Pane: SQNR
ax_sqnr.set_title("System Quality: SQNR (dB)", fontsize=13, fontweight='bold')
ax_sqnr.set_ylabel("dB")
ax_sqnr.grid(True, which="both", ls="-", alpha=0.2)
ax_sqnr.legend(loc='upper left', bbox_to_anchor=(1.02, 1), title="Scenarios")

# Formatting Middle Pane: Max Error
ax_err.set_title("Bit-Level Integrity: Maximum Absolute Error", fontsize=13, fontweight='bold')
ax_err.set_ylabel("Absolute Error")
ax_err.set_yscale('log')
ax_err.grid(True, which="both", ls="-", alpha=0.2)

# Formatting Bottom Pane: Block Exponent
ax_exp.set_title("Hardware Scaling Activity: Block Exponent", fontsize=13, fontweight='bold')
ax_exp.set_ylabel("Shifts (Stages)")
ax_exp.set_xlabel("Sample Size (N)")
ax_exp.set_xscale('log', base=2)
ax_exp.set_xticks(N_values)
ax_exp.get_xaxis().set_major_formatter(plt.ScalarFormatter())
ax_exp.grid(True, which="both", ls="-", alpha=0.2)

# --- Improved Clock Cycle Analysis ---
ax_cc.set_title("Computational Efficiency: Clock Cycles vs. Complexity", fontsize=13, fontweight='bold')

# 1. Plot the actual hardware cycles
ax_cc.plot(data['N'], data['cycles'], marker='D', color='forestgreen', linewidth=3, label="Actual HW Cycles")

# 2. Add a 'Theoretical' line for O(N log2 N)
# We scale it to match your N=16 point to show the trend
n_log_n = np.array(N_values) * np.log2(N_values)
scaling_factor = data['cycles'][0] / n_log_n[0]
ax_cc.plot(N_values, n_log_n * scaling_factor, linestyle=':', color='black', alpha=0.5, label="O(N log2 N) Trend")

# 3. Annotations for Latency
for i, txt in enumerate(data['cycles']):
    ax_cc.annotate(f"{int(txt)}", (data['N'][i], data['cycles'][i]), 
                   textcoords="offset points", xytext=(0,10), ha='center', fontsize=9)

ax_cc.set_ylabel("Clock Cycles (log scale)")
ax_cc.set_yscale('log')
ax_cc.set_xscale('log', base=2)
ax_cc.set_xticks(N_values)
ax_cc.get_xaxis().set_major_formatter(plt.ScalarFormatter())

# Add a twin axis for Time (Assuming a 100MHz Clock)
ax_time = ax_cc.twinx()
ax_time.set_ylabel("Time @ 100MHz (µs)")
ax_time.set_yscale('log')
# Convert cycles to microseconds: (cycles / 100,000,000) * 1,000,000
ax_time.set_ylim(ax_cc.get_ylim()[0]/100, ax_cc.get_ylim()[1]/100) 

ax_cc.grid(True, which="both", ls="-", alpha=0.2)
ax_cc.legend(loc='upper left')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.utils import capture
from fft_utils import run_sim, analyze_fft_performance

# --- IFFT Round-Trip Verification Cell ---
N_rt = 1024
t = np.arange(N_rt) / N_rt

# 1. Create a stimulus with "Real World" complexity
# A mix of two sines with an exponential decay
stimulus = (0.8 * np.exp(-2*t) * (np.sin(2*np.pi*12.3*t) + 0.5*np.cos(2*np.pi*40.5*t))).astype(complex)


print(f"🔄 Starting IFFT Round-Trip (N={N_rt})...")

# 2. Forward FFT Stage
hw_fft, exp_fft, _ = run_sim(stimulus, invert=False)
# Reconstruct hardware result to feed into next stage
fft_mid_signal = hw_fft #* (2.0**exp_fft)

# 3. Inverse FFT Stage
hw_ifft, exp_ifft, _ = run_sim(fft_mid_signal, invert=True)
reconstructed = hw_ifft * (2**(exp_ifft+exp_fft)) / N_rt

# 4. Final Comparison
# We compare the 'reconstructed' output against the 'stimulus' input
error_rt = stimulus - reconstructed
mse_rt = np.mean(np.abs(error_rt)**2)
signal_pwr_rt = np.mean(np.abs(stimulus)**2)
sqnr_rt = 10 * np.log10(signal_pwr_rt / mse_rt)
enob_rt = (sqnr_rt - 1.76) / 6.02

print("-" * 50)
print(f"✅ RECONSTRUCTION COMPLETE")
print(f"Round-Trip SQNR: {sqnr_rt:.2f} dB")
print(f"Effective Precision: {enob_rt:.2f} bits")
print("-" * 50)

# Plotting the results
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))
ax1.plot(stimulus.real, label="Original (Real)", color='gray', alpha=0.5, lw=3)
ax1.plot(reconstructed.real, label="Reconstructed (Real)", color='dodgerblue', linestyle='--')
ax1.set_title("Time Domain: Original vs. Hardware Round-Trip")
ax1.legend()

ax2.plot(20*np.log10(np.abs(error_rt) + 1e-12), color='crimson', lw=1)
ax2.set_title("Reconstruction Error Floor (dB)")
ax2.set_ylabel("Error Magnitude")
plt.tight_layout()
plt.show()
```
